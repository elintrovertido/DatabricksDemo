```sql
%sql
CREATE TABLE EMPLOYEES(
ID INT ,
NAME STRING,
SALARY DOUBLE
) 
USING DELTA;
```

```sql
%sql
INSERT INTO EMPLOYEES VALUES (1, 'SANTHOSH', 1000);
INSERT INTO EMPLOYEES VALUES (2, 'SHIVA', 2000);
INSERT INTO EMPLOYEES VALUES (3, 'SUMANTH', 3000);
```

```sql
%sql
DESCRIBE DETAIL EMPLOYEES;
```

```sql
%sql
UPDATE employees
SET SALARY  = SALARY + 100
WHERE SALARY < 2000;
```
